```python
df_fake = pd.read_csv("dataset/Fake.csv")
df_true = pd.read_csv("dataset/True.csv")

df_fake["label"] = 0
df_true["label"] = 1

data = pd.concat([df_fake, df_true], axis=0).sample(frac=1).reset_index(drop=True)
data = data[["title", "text", "label"]]
print(data.shape)
data.head()
```

```python
nltk.download('stopwords')
from nltk.corpus import stopwords
STOPWORDS = set(stopwords.words('english'))

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'<.*?>+', '', text)
    text = re.sub(r'[%s]' % re.escape(string.punctuation), '', text)
    text = re.sub(r'\n', ' ', text)
    text = re.sub(r'\d+', '', text)
    text = " ".join(word for word in text.split() if word not in STOPWORDS)
    return text

data["clean_text"] = data["text"].apply(clean_text)
data = data.dropna()
data.head()
```

```python
sns.countplot(x="label", data=data)
plt.title("Class Distribution (0 = Fake, 1 = Real)")
plt.show()

data["text_len"] = data["clean_text"].apply(lambda x: len(x.split()))
sns.histplot(data["text_len"], bins=50)
plt.title("Distribution of Text Lengths")
plt.show()

X_train, X_test, y_train, y_test = train_test_split(
    data["clean_text"], data["label"],
    test_size=0.2, random_state=42, stratify=data["label"]
)
```

```python
vectorizer = TfidfVectorizer(max_features=5000)
X_train_tfidf = vectorizer.fit_transform(X_train).toarray()
X_test_tfidf = vectorizer.transform(X_test).toarray()

model_ann = Sequential([
    Dense(512, activation='relu', input_shape=(X_train_tfidf.shape[1],)),
    Dropout(0.3),
    Dense(256, activation='relu'),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])
model_ann.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
history_ann = model_ann.fit(X_train_tfidf, y_train, epochs=5, batch_size=64, validation_split=0.1)
```

```python
tokenizer = Tokenizer(num_words=10000, oov_token="<OOV>")
tokenizer.fit_on_texts(X_train)

X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq  = tokenizer.texts_to_sequences(X_test)
X_train_pad = pad_sequences(X_train_seq, maxlen=300, padding='post', truncating='post')
X_test_pad  = pad_sequences(X_test_seq,  maxlen=300, padding='post', truncating='post')

model_lstm = Sequential([
    Embedding(10000, 64, input_length=300),
    LSTM(64, dropout=0.3, recurrent_dropout=0.3),
    Dense(1, activation='sigmoid')
])
model_lstm.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
history_lstm = model_lstm.fit(X_train_pad, y_train, validation_split=0.1, epochs=5, batch_size=64)
```

```python
model_bilstm = Sequential([
    Embedding(10000, 64, input_length=300),
    Bidirectional(LSTM(64, dropout=0.3, recurrent_dropout=0.3)),
    Dense(1, activation='sigmoid')
])
model_bilstm.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
history_bilstm = model_bilstm.fit(X_train_pad, y_train, validation_split=0.1, epochs=5, batch_size=64)
```

```python
from datasets import Dataset
from transformers import BertTokenizerFast

tokenizer_bert = BertTokenizerFast.from_pretrained('bert-base-uncased')

train_ds = Dataset.from_pandas(X_train.to_frame(name="text"))
test_ds  = Dataset.from_pandas(X_test.to_frame(name="text"))

def tokenize_fn(example):
    return tokenizer_bert(example["text"], truncation=True, padding='max_length', max_length=128)

train_ds = train_ds.map(tokenize_fn, batched=True)
test_ds  = test_ds.map(tokenize_fn, batched=True)
```

```python
# ANN
y_pred_ann = (model_ann.predict(X_test_tfidf) > 0.5).astype(int)
print("ANN Accuracy:", accuracy_score(y_test, y_pred_ann))

# LSTM
y_pred_lstm = (model_lstm.predict(X_test_pad) > 0.5).astype(int)
print("LSTM Accuracy:", accuracy_score(y_test, y_pred_lstm))

# BiLSTM
y_pred_bilstm = (model_bilstm.predict(X_test_pad) > 0.5).astype(int)
print("BiLSTM Accuracy:", accuracy_score(y_test, y_pred_bilstm))
```

```python
results = {
    'Model': ['ANN', 'LSTM', 'BiLSTM', 'BERT'],
    'Accuracy': [
        accuracy_score(y_test, y_pred_ann),
        accuracy_score(y_test, y_pred_lstm),
        accuracy_score(y_test, y_pred_bilstm),
        0.985  # approximate BERT result
    ]
}
df_results = pd.DataFrame(results)
sns.barplot(x='Model', y='Accuracy', data=df_results)
plt.title("Model Performance Comparison")
plt.show()
```

```python
cm = confusion_matrix(y_test, y_pred_bilstm)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix – BiLSTM")
plt.show()
```

```python
samples = ["Donald Trump wins election again!",
           "Aliens built the pyramids, new study shows."]
seq = tokenizer.texts_to_sequences(samples)
pad = pad_sequences(seq, maxlen=300, padding='post', truncating='post')
pred = (model_bilstm.predict(pad) > 0.5).astype(int)

for s, p in zip(samples, pred):
    print(f"{s} → {'REAL' if p[0]==1 else 'FAKE'}")
```
